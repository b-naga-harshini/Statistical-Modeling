## Selecting the Predictor and Target Variables

```python
feature = ["Price","Number of products sold","Stock levels","Lead times","Order quantities","Shipping times","Shipping costs","Production volumes","Manufacturing lead time","Manufacturing costs","Defect rates","Costs"]
target = "Revenue generated"
```

## Creating X and Y values

```python
x = df[feature]
y = df[target]
```

## train_test_split

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state=42)
```

## Create Model

```python
model = LinearRegression()
```

## train the model and generated prediction

```python
model.fit(x_train,y_train)
```

```python
y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)
```

## Training Performance

```python
R2 = r2_score(y_train, y_train_pred)
print("\nR2 Score:",R2)
```

## Testing Performance

```python
r2 = r2_score(y_test,y_test_pred)

mse = mean_squared_error(y_test,y_test_pred)

mae = mean_absolute_error(y_test,y_test_pred)

rmse = np.sqrt(mse)

print("\n========Test Result=======")

print("R² Score:", r2)

print("Mean Absolute Error (MAE):", mae)

print("Mean Squared Error (MSE):", mse)

print("Root Mean Squared Error (RMSE):", rmse)
```
